````python
# 習題：數學物件、Python 型別與 AI 中的數值表示

在數學中，我們經常接觸下列概念：

- 自然數（Natural Numbers）
- 整數（Integers）
- 有理數（Rational Numbers）
- 實數（Real Numbers）
- 複數（Complex Numbers）
- 向量（Vectors）
- 矩陣（Matrices）
- 集合（Sets）
- 命題的真假（Truth Values）

本習題將探討：

```text
數學物件
    ↓
Python 型別
    ↓
CPython 實作
    ↓
AI 中的數值表示
```

---

## 第一部分：Python 如何表示數學物件？

請查閱 Python 文件或自行測試，完成下表。

| 數學概念 | Python 型別 | 範例 |
|----------|------------|------|
| 自然數 | int | 5 |
| 整數 | int| -10 |
| 有理數 | float | 0.75 |
| 實數 | float | 3.14 |
| 複數 |complex | 2+3j |
| 向量 | list | [1, 2, 3] |
| 矩陣 | list | [[1, 2], [3, 4]] |
| 集合 | set | {1, 2, 3} |
| 命題真假 | bool | True |

並回答下列問題：

1. 哪些是 Python 的內建型別（Built-in Types）？Python 的內建型別包含 int（整數）、float（浮點數）、complex（複數）、bool（布林值）、list（串列）以及 set（集合）。這些型別不需要額外安裝任何模組即可直接使用。

2. 哪些需要額外模組才能較完整地表示？向量與矩陣雖然可以使用 list 來表示，但若要進行較完整且有效率的數學運算，通常會使用額外的模組，例如 NumPy。

3. Python 的型別分類與數學中的分類是否完全一致？並不完全一致。數學是依照概念分類，例如自然數、整數、有理數和實數；而 Python 是依照資料儲存方式分類。例如自然數和整數在數學上是不同概念，但在 Python 中都使用 int 型別來表示。

---

## 第二部分：數學概念與 Python 型別

數學上有下列包含關係：

$$
\mathbb N
\subseteq
\mathbb Z
\subseteq
\mathbb Q
\subseteq
\mathbb R
\subseteq
\mathbb C
$$

請回答：

1. Python 是否為上述每一個數系都提供獨立型別？不是。Python 並沒有為每一個數學數系都提供獨立型別，例如自然數和整數都使用 int，有理數和實數通常都使用 float 表示。

2. 為什麼自然數與整數都使用 `int` 表示？因為在電腦中，自然數和整數都是沒有小數點的數字，Python 不特別區分兩者，因此統一使用 int 型別來表示。

3. Python 為什麼提供 `complex` 型別？因為複數是數學與工程領域常用的數值，因此 Python 內建 complex 型別，方便直接進行複數運算。

4. Python 為什麼沒有內建的有理數型別？
因為大部分數值運算使用 float 已經足夠，因此 Python 預設沒有內建專門的有理數型別。若需要精確表示分數，可以使用額外模組來處理。

5. 下列三個物件的型別有何不同？數值雖然都表示 1，但型別不同。1 的型別是 int（整數），1.0 的型別是 float（浮點數、小數），1+0j 的型別是 complex（複數）。Python 會根據數值的表示方式判斷其型別，因此即使它們代表的數值相同，仍然屬於不同的資料型別。

```python
1
1.0
1+0j
```

---

## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 可表示任意大小的整數 |
| 整數 | `int` | PyLongObject | 支援正整數與負整數 |
| 實數 | `float` |PyFloatObject | 採用浮點數表示，可能有誤差 |
| 複數 | `complex` | PyComplexObject | 可直接進行複數運算 |
| 集合 | `set` | PySetObject | 元素不重複、無固定順序 |
| 命題真假 | `bool` | PyBoolObject | 只有 True 與 False |
| 串列 | `list` | PyListObject | 可儲存多個資料且可修改 |

---

## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？是，int 是 Python 的內建型別。

2. 對應的 CPython 物件結構名稱為何？
是，int 是 Python 的內建型別。

3. 它如何儲存資料？以整數形式儲存資料。

4. 是否有固定大小限制？沒有固定大小限制。

5. 若資料規模持續增加，如何擴充儲存空間？當數值變大時，Python 會自動配置更多記憶體來儲存。

---

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為 Python 的 int 採用任意精度（Arbitrary Precision）設計，可以根據需要自動增加儲存空間，因此能表示非常大的整數。

2. Python 的 `int` 是否具有固定的位元數限制？沒有。Python 的 int 不像許多程式語言只有 32 位元或 64 位元的限制，它會依照數值大小動態擴充。

3. 什麼是 Arbitrary Precision（任意精度）？任意精度是指數字的大小不受固定位元數限制。當數值持續變大時，Python 會自動配置更多記憶體來儲存，因此可以處理非常大的整數。

4. CPython 使用什麼物件來實作 `int`？CPython 使用 PyLongObject 來實作 int 型別。

5. Python 整數大小真正的限制是什麼？理論上 Python 的整數沒有固定大小限制，真正的限制來自電腦可用的記憶體。如果記憶體不足，就無法再繼續儲存更大的整數。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？因為電腦使用二進位來儲存浮點數，而有些十進位小數無法被二進位精確表示，所以計算結果可能出現微小誤差。

2. 浮點數是否能精確表示所有小數？不能。有些小數可以精確表示，但像 0.1、0.2 這類數字無法在二進位中完全表示，因此會產生誤差。

3. 什麼是浮點數的精度（precision）？精度是指浮點數能夠準確表示的有效數字位數。Python 的 float 通常可提供約 15 到 17 位有效數字的精度。

4. 什麼是浮點數的範圍（range）？範圍是指浮點數能表示的數值大小區間，也就是從非常小的數到非常大的數都能儲存，但超出範圍可能會發生溢位或下溢。

5. 精度與範圍有何不同？精度表示數字有多精確，範圍表示數字能有多大或多小。精度關心的是小數點後能準確到多少位，範圍關心的是可表示數值的大小限制。
---

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？AI 模型中的權重通常使用浮點數（Floating Point）儲存，例如 FP32、FP16 或 BF16，因為浮點數能夠表示小數，適合神經網路的運算需求。

2. 什麼是 FP64？FP64（64-bit Floating Point）是 64 位元浮點數，也稱為雙精度浮點數（Double Precision），具有較高的精度，但需要較多的記憶體空間。

3. 什麼是 FP32？FP32（32-bit Floating Point）是 32 位元浮點數，也稱為單精度浮點數（Single Precision），是在深度學習中常見的數值表示方式。

4. 什麼是 FP16？FP16（16-bit Floating Point）是 16 位元浮點數，所需記憶體較少，運算速度較快，因此常用於大型 AI 模型的訓練與推論。

5. 什麼是 BF16？BF16（BFloat16）是一種 16 位元浮點數格式，保留與 FP32 相近的數值範圍，同時減少記憶體使用量，因此廣泛應用於現代 AI 系統。

6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？因為 AI 模型的權重通常需要表示小數，而 int 只能表示整數，無法精確描述神經網路中的權重變化。此外，浮點數在矩陣運算與機器學習運算上效率更高，因此大型 AI 模型通常使用 FP32、FP16 或 BF16 作為權重的儲存格式。

---

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int |不固定 | 否 | 儲存整數 |
| FP64 | 64 位元 | 有 | 科學計算、高精度運算 |
| FP32 | 32 位元 | 有 | AI 訓練、一般浮點運算 |
| FP16 | 16 位元 | 有 | AI 訓練與推論，節省記憶體|
| BF16 | 16 位元 | 有 | 現代 AI 訓練與推論 |

並回答：

1. 哪種表示方式最精確？在表中的型別中，FP64 最精確，因為它使用 64 位元儲存資料，能保留較多有效數字。

2. 哪種表示方式最節省記憶體？FP16 和 BF16 最節省記憶體，因為它們只使用 16 位元儲存資料。

3. 哪種表示方式最常見於 AI 訓練？目前 AI 訓練常使用 FP32、FP16 以及 BF16，其中 BF16 和 FP16 因為能降低記憶體需求而越來越普遍。

4. 哪種表示方式最常見於 AI 推論（Inference）？AI 推論最常使用 FP16 或 BF16，因為可以減少記憶體使用量並提升運算速度。

---

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？Unlimited Precision（無限制精度）是指 Python 的 int 不受固定 32 位元或 64 位元限制，能夠依照需要儲存非常大的整數。

2. 它是否代表可以儲存無限大的整數？不完全是。雖然 Python 的 int 沒有固定大小限制，但仍受到電腦記憶體大小的限制，因此無法真正儲存無限大的整數。

3. CPython 如何實現這項特性？CPython 使用 PyLongObject 來實作 int 型別。當整數變大時，PyLongObject 會自動配置更多記憶體來儲存額外的數字，因此能夠表示遠大於一般 32 位元或 64 位元範圍的整數。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？因為 Python 的 int 採用任意精度（Arbitrary Precision）設計，當整數變大時會自動使用更多記憶體儲存，因此不像一般程式語言受限於固定的 32 位元或 64 位元大小。

2. 為什麼 `float` 卻存在範圍限制？因為 float 採用 IEEE 754 浮點數標準，以固定的位元數來儲存資料，所以能表示的數值範圍有限，超出範圍可能會造成溢位。

3. 為什麼 `float` 可能產生誤差？因為電腦使用二進位儲存浮點數，而某些十進位小數無法被二進位精確表示，例如 0.1，因此在運算時可能出現微小誤差。

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Python 的 int 採用 Arbitrary Precision（任意精度）設計，重視計算結果的正確性，因此可以表示非常大的整數。AI 模型常用的 FP16 與 BF16 則採用 Limited Precision（有限精度）設計，重視運算速度與記憶體效率，因此允許一定程度的誤差。

2. 各自的優點與缺點。Python 的 int 優點是能精確表示大整數且不易發生溢位，缺點是需要較多記憶體與運算資源。FP16 與 BF16 的優點是佔用較少記憶體、運算速度較快，缺點是精度較低，可能產生數值誤差。

3. 為什麼大型語言模型通常選擇較低精度的表示方式？因為大型語言模型需要儲存大量參數並進行大量運算。使用 FP16 或 BF16 可以大幅減少記憶體需求，提升訓練與推論速度，而模型效能通常不會受到太大影響，因此較具效率。

4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？記憶體使用量會大幅增加，運算速度也可能變慢，需要更高規格的硬體設備，導致訓練成本與推論成本上升，因此較不適合大型 AI 模型的實際應用。
---

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

答:Python 可以正常計算 10**1000，是因為它的整數會自動增加儲存空間，所以能處理非常大的數字。不過大型 AI 模型需要儲存幾十億甚至幾千億個參數，如果全部都用高精度來計算，會非常占記憶體，速度也會變慢。因此 AI 通常使用 FP32、FP16 或 BF16，雖然精度比較低，可能有些微誤差，但能節省大量記憶體、提高 GPU 運算效率，讓訓練和推論速度更快。對 AI 來說，速度和效率比完全精確更重要，所以較常使用較低精度的數值格式。
````
